# 1. Load and Prepare Data

```python
frequent_itemsets=pd.read_csv('FR_ITEMSETS.CSV')
frequent_itemsets
```

The additional column ('Unnamed') is not really needed here

And there is an even bigger problem, if we look at the data type of column "itemsets":

```python
itemsets=frequent_itemsets['itemsets']
type(itemsets[0])
```

We cannot build association rules from string objects. Therefore, we need a little trick

```python
frequent_itemsets=pd.read_csv('FR_ITEMSETS.CSV',header=0, index_col=0, converters={'itemsets': ast.literal_eval})
frequent_itemsets
```

Now this looks better:

```python
itemsets=frequent_itemsets['itemsets']
type(itemsets[0])
```
