# Sampling

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets

#load iris data set
x = datasets.load_iris(return_X_y=False)
df=pd.DataFrame(x.data,columns=x.feature_names)
df['label']=x.target
df
```

# Simple random sampling with replacement

```python
from sklearn.utils import resample

df1 = resample(df, n_samples=10)
df2 = resample(df, n_samples=10)
```

# Balanced Sampling

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load the iris dataset
iris=pd.DataFrame(x.data,columns=x.feature_names)
iris['label']=x.target
#iris

non_balanced_iris = iris.iloc[0:110,:]
non_balanced_iris['label'].value_counts()
```

```python
#!pip install imblearn
```

```python
#split data to features and label
X= non_balanced_iris.iloc[:,:4]
y= non_balanced_iris['label']
```

```python
from imblearn.over_sampling import RandomOverSampler

# Create an instance of RandomOverSampler
oversampler = RandomOverSampler(random_state=0)

# Fit and apply the oversampling to the data
X_resampled, y_resampled = oversampler.fit_resample(X, y)

# Print the original and resampled class distributions
print('Original class distribution:', {i:sum(y==i) for i in range(3)})
print('Resampled class distribution:', {i:sum(y_resampled==i) for i in range(3)})
```

```python
balanced_iris=pd.DataFrame(X_resampled,columns=non_balanced_iris.columns)
balanced_iris['label']=y_resampled
balanced_iris[balanced_iris['label'] == 2].head(20)
```

# Stratified Sampling

```python
df
```

```python
#Variation Stratified Sampling: Enforces a representative sample concerning classes
#Example1: Already balanced data set
df_train, df_test = train_test_split(df, test_size=0.4, random_state=1,stratify=df['label'])

print("Shape of training sample: ",df_train.shape)
print("Shape of testing sample: ",df_test.shape)

print("\nDistribution of labels in training sample\n",df_train['label'].value_counts())
print("Distribution of labels in test sample\n",df_test['label'].value_counts())
```

```python
#Example2: Already imbalanced data set
df_train, df_test = train_test_split(non_balanced_iris, test_size=0.4, random_state=1,stratify=non_balanced_iris['label'])

print("Shape of training sample: ",df_train.shape)
print("Shape of testing sample: ",df_test.shape)

print("\nDistribution of labels in training sample\n",df_train['label'].value_counts())
print("Distribution of labels in test sample\n",df_test['label'].value_counts())
```

# Split into training - validation - test data

```python
#Variation Train-Validation-Test Sampling
df_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_train, test_size=0.25, random_state=1)

print("Shape of training sample: ",df_train.shape)
print("Shape of validation sample: ",df_val.shape)
print("Shape of testing sample: ",df_test.shape)
```

# Cross Validation

In Cross Validation, the original data set is split multiple times into training and test data sets.

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=5)
skf.get_n_splits(df)

print(skf)

for i, (train_index, test_index) in enumerate(skf.split(df.iloc[:,:4],df['label'])):
    print("\n")
    print(f"Fold {i}:")
    print(f"  Train: index={train_index}")
    print(f"  Test:  index={test_index}")
```
