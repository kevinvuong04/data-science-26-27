The edit distance between two strings is the minimum number of string operations (or edit operations) needed to convert one string in the other

Three types of edit operations are usually considered:
- inserting a character
- replacing a character or 
- deleting a character


```python
#!pip install editdistance
```

```python
#Edit Distance
import editdistance

editdistance.eval('banana', 'bahama')
```

| Input | Output | Operation |
| --- | --- | --- |
| Commputa | Computa | Delete „m“ |
| Computa | Compute | Replace „a“ -> „e“ |
| Compute | Computer | Insert „r“ |
		

```python
editdistance.eval('Commputa', 'Computer')
```

Jaro–Winkler distance is a string metric measuring the edit distance between two sequences in a normalized score. 
The lower the score the more similar the strings are: 1 means an exact match and 0 means there is no similarity.

```python
#!pip install pyjarowinkler
```

```python
from pyjarowinkler import distance
# Scaling is 0.1 by default

#distance.get_jaro_distance("ALGORITHM", "LOGARITHMUS", winkler=True, scaling=0.1))

print ("Jaro-Winkler distance: ",distance.get_jaro_distance("ALGORITHM", "LOGARITHMUS", winkler=True, scaling=0.1))
print ("Jaro distance: ",distance.get_jaro_distance("ALGORITHM", "LOGARITHMUS", winkler=False))
```
