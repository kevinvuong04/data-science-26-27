```python
####################### Alternative function to calculate and print p-values without statsmodels package #######################

def calc_pvalues(X, y, ypred,reg_model):
    # Predictions and residuals
    residuals = y - ypred

    # Degrees of freedom
    n, k = X.shape
    df = n - k

    # Residual variance
    residual_var = np.sum(residuals ** 2) / df

    # Add intercept term if model includes intercept
    if reg_model.fit_intercept:
        X_design = np.column_stack((np.ones(n), X))
        coef = np.insert(reg_model.coef_, 0, reg_model.intercept_)
        feature_names = ['Intercept'] + list(X.columns)
    else:
        X_design = X
        coef = reg_model.coef_

    # Variance-Covariance matrix of beta
    XTX_inv = np.linalg.inv(X.T @ X)
    var_beta = residual_var * XTX_inv

    # Standard errors of coefficients
    se_beta = np.sqrt(np.diag(var_beta))

    # t-statistics
    t_stats = coef[1:] / se_beta

    # p-values from t-distribution (two-tailed)
    p_values = 2 * (1 - stats.t.cdf(np.abs(t_stats), df))
   
    # Display the results
    #drop first feature name (intercept)
    results = pd.DataFrame({
        'Feature': feature_names[1:],
        'Coefficient': coef[1:],
        'Standard Error': se_beta,
        't-Statistic': t_stats,
        'p-Value': p_values
    })
    print(results)
```

# Task 4: Algorithm Selection and Model Tuning

```python
#Build KNN Regression Model using the training data and standard parametrization
from sklearn.neighbors import KNeighborsRegressor
KNN_reg = KNeighborsRegressor()

#Fit regression model to training data
KNN_reg.fit(train_features,train_labels)
```

```python
#Predict all test examples
ypred = KNN_reg.predict(test_features)

#Evaluate KNN regression model with test data using Mean Absolute Error
mae = mean_absolute_error(test_labels, ypred)
print('MAE: %.3f' % mae)
```

Build a function, which receives the number of neighbors as an argument!

The function should 
- set the KNN regressor to the defined number of neighbors,
- build a model with this setting using the training data, 
- predict values for the test data and 
- calculate and print the MAE

```python
def tune_and_evaluate_knn_reg(number_of_neighbors):
    #Use the method set_params to set the number of neighbors
    KNN_reg.set_params(n_neighbors=number_of_neighbors)
    #Fit regression model to training data
    KNN_reg.fit(train_features,train_labels)
    #Predict all test examples
    ypred = KNN_reg.predict(test_features)
    #Evaluate KNN regression model with test data using Mean Absolute Error
    mae = mean_absolute_error(test_labels, ypred)
    print('MAE: %.3f' % mae)
```

```python
#call your funtion with number of neighbors = 10
tune_and_evaluate_knn_reg(10)
```
