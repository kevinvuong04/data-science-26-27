# 1. Load and Prepare data

```python
#Load transaction data (there are no headers)
filename='Market_Basket_Optimisation.csv'
with open(filename) as file_object:
    basket_data = file_object.readlines()
```

```python
#Transform market basket data into list-in-a-list (all Items with capitalized first letters)
transactions=[]
for transaction in basket_data:
    #remove line breaks
    transaction_clean=transaction.rstrip()
    #split string into list
    transaction_list=transaction_clean.split(",")
    #capitalize first letters
    transactions.append([item.title() for item in transaction_list])
    
print("Example transaction: ",transactions[0])
```

# 4. Export to CSV

Unfortunately itemsets are stored as frozensets by default, which is a bit uncomfortable to import in step 2. Therefore, we transform them from frozenset => set before export:

```python
new_itemsets=[]
itemsets=frequent_itemsets['itemsets']
for x in itemsets:
    new_itemsets.append(set(x))

frequent_itemsets['itemsets']=new_itemsets
```
