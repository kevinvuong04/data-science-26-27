# Logistic regression and k Nearest Neighbors Demo

```python
#load data set
from sklearn import neighbors, datasets
iris = datasets.load_iris()

x, y = iris.data, iris.target

print("Format x: ",type(x), x.shape)
print("Format y: ",type(y), y.shape)
```

```python
#split part of the data for training
x_train=x[:120,]
y_train=y[:120,]
```

```python
from sklearn.linear_model import LogisticRegression

#Initialize Logistic regression model
lr = LogisticRegression(random_state=0,max_iter=10000)

#fit model to training data
lr.fit(x_train,y_train)
#predict label for a given data set
print("Predicted label for iris flower with feature values [3, 5, 4, 2]: ",iris.target_names[lr.predict([[3, 5, 4, 2]])])
```

```python
#split part of the data for testing
x_test=x[120:150,]
y_test=y[120:150,]

#Predict multiple values (e.g. predictions for entire test data set)
lr.predict(x_test)
```

```python
#calculate accuracy for test data
lr.score(x_test,y_test)
```

```python
#k Nearest Neighbors Algorithm
from sklearn import neighbors

knn = neighbors.KNeighborsClassifier(n_neighbors=10)
knn.fit(x_train,y_train)

knn.score(x_test,y_test)
```
